```python
Score = [88,93,-5,76,105,84,99,67]
valid_scores = []
invalid_scores = []
for Score in Score:
    if Score >= 0 and Score <=100:
        valid_scores.append(Score)
    else:
        invalid_scores.append(Score)
print(f"valid_scores: {valid_scores}")
print(f"invalid_scores: {invalid_scores}")


        
```

Invalid values should be identified before calculating statistics because they can make the results incorrect. 

```python
def calculate_average(values):
    total = 0
    for value in values:
        total = total + value
    average = total / len(values) 
    return average


def classify_score(Scores):
    if Scores >= 90:
        return "A"
    elif Scores >= 80:
        return "B"
    elif Scores >= 70:
        return "C"
    elif Scores >= 60:
        return "D"
    else:
        return "F"
average = calculate_average(valid_scores)
print(f"Avarage: {average}")
for scores in valid_scores:
    grade = classify_score(scores)
    print(f"{scores} -> {grade}")


        

```

One advantage of using a function is that it makes the code reusable. Instead of writing the same code again every time, you can simply call the function. This makes the program shorter, easier to understand, and easier to update.

```python

name = [ 
    "  alice johnson ", 
    "BOB SMITH", 
    "  carol WILLIAMS",     "david brown  " 
    ] 

def clean_name(name):
    name = name.strip()
    name = name.title()
    return name
clean_names = []

for name in name:
  clean_names.append(clean_name(name))

for name in clean_names:
   parts = name.split()
   first_name = parts[0]

   print(f"{name} ->  first_name: {first_name}")
```

Consistent text formatting is important because it helps Python recognize matching data correctly. 

```python
students = [ 
    {"name": "Alice", "major": "Computer Science", "score": 92},     {"name": "Bob", "major": "Business", "score": 81}, 
    {"name": "Carol", "major": "Computer Science", "score": 88}, 
    {"name": "David", "major": "Biology", "score": 76}, 
    {"name": "Emma", "major": "Computer Science", "score": 95}
    ]

for student in students:
    print(student["name"] , student["major"] , student["score"])

cs_students = [] 

for student in students:
  if student ["major"] == "Computer Science":
    cs_students.append(student)
            
for student in cs_students:
   print(student["name"] , student["score"])   

total = 0
for student in cs_students: 
   total = total + student["score"]    
average = total /len(cs_students) 
print(average) 

highest_student = students[0]
for student in students:
   if student["score"] > highest_student["score"]:
      highest_student = student
print(f"Highest Score: {highest_student['name']} with {highest_student['score']}")      
        
```

The dataset shows that Computer Science students generally performed well compared with the other majors. I also learned how Python can be used to filter students by major, calculate an average, and identify patterns in student performance.
